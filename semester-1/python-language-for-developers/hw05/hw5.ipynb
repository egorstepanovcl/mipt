# Задача 1

Напишите программу, которая создаёт 2 потока для вычисления квадратов и кубов целых чисел от 1 до 10.

```python
import threading

def squares():
    for i in range(1, 11):
        print(f"Квадрат числа {i} = {i ** 2}")

def cubes():
    for i in range(1, 11):
        print(f"Куб числа {i} = {i ** 3}")

thread1 = threading.Thread(target=squares)
thread2 = threading.Thread(target=cubes)

thread1.start()
thread2.start()

thread1.join()
thread2.join()

print("Вычисления завершены.")
```

# Задача 2

Напишите программу, которая создаёт несколько потоков для выполнения функции, которая выводит целые числа от 1 до 10 с задержкой в 1 секунду.

```python
import threading
import time

def print_numbers(name):
    for i in range(1, 11):
        print(f"{name}: {i}")
        time.sleep(1)

threads = []

for n in range(3):
    t = threading.Thread(target=print_numbers, args=(f"Поток {n+1}",))
    threads.append(t)
    t.start()

for t in threads:
    t.join()

print("Все потоки завершились.")
```
