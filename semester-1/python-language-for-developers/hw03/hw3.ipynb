# Задача «Модель магазина»

Создайте модель для онлайн-магазина, который управляет товарами и заказами. Ваша задача — реализовать классы для товаров, заказов и магазина, который будет обрабатывать заказы.

## Требования

1. Класс `Product`

    Атрибуты:
      - `name` (строка) — название товара;
      - `price` (число) — цена товара;
      - `stock` (целое число) — количество товара на складе.

    Методы:
      - `update_stock(quantity)` — обновляет количество товара на складе; при уходе в отрицательное значение должно выводиться сообщение об ошибке.

2. Класс `Order`

    Атрибуты:
      - `products` (словарь) — ключ: объект `Product`, значение: количество этого товара в заказе.

    Методы:
      - `add_product(product, quantity)` — добавляет товар в заказ; если на складе недостаточно, выводится сообщение об ошибке;
      - `calculate_total()` — рассчитывает общую стоимость заказа.

3. Класс `Store`

    Атрибуты:
      - `products` (список) — список всех доступных товаров в магазине.

    Методы:
      - `add_product(product)` — добавляет товар в магазин;
      - `list_products()` — отображает все товары с их ценами и остатками;
      - `create_order()` — создаёт новый заказ.

### Дополнительные задания

1. Удаление товаров из заказа

    Добавьте метод `remove_product(product, quantity)` в класс `Order`, который уменьшает количество указанного товара; если количество становится нулём, удалите товар из словаря `products`.
2. Возвраты и обновление запасов

    Реализуйте метод `return_product(product, quantity)` в классе `Order`, который возвращает товар в магазин: уменьшает количество в заказе, вызывает `product.update_stock(quantity)` для увеличения остатка, и удаляет позицию из `products`, если её количество стало нулём.

```python
class Product:
    def __init__(self, name: str, price: float, stock: int):
        self.name = name
        self.price = price
        self.stock = stock

    def update_stock(self, quantity: int):
        new_stock = self.stock + quantity
        if new_stock < 0:
            print(f"Ошибка: недостаточно товара '{self.name}' на складе для обновления.")
        else:
            self.stock = new_stock

class Order:
    def __init__(self):
        self.products = {}

    def add_product(self, product: Product, quantity: int):
        if quantity <= 0:
            print("Количество для добавления должно быть положительным.")
            return
        if product.stock < quantity:
            print(f"Ошибка: недостаточно товара '{product.name}' на складе для заказа.")
            return
        product.update_stock(-quantity)
        if product in self.products:
            self.products[product] += quantity
        else:
            self.products[product] = quantity

    def remove_product(self, product: Product, quantity: int):
        if product not in self.products:
            print(f"Товар '{product.name}' отсутствует в заказе.")
            return
        if quantity <= 0:
            print("Количество для удаления должно быть положительным.")
            return
        if self.products[product] < quantity:
            print(f"Ошибка: в заказе меньше товара '{product.name}', чем пытаются удалить.")
            return
        self.products[product] -= quantity
        if self.products[product] == 0:
            del self.products[product]

    def calculate_total(self):
        total = 0
        for product, quantity in self.products.items():
            total += product.price * quantity
        return total

    def return_product(self, product: Product, quantity: int):
        if product not in self.products:
            print(f"Товар '{product.name}' отсутствует в заказе для возврата.")
            return
        if quantity <= 0:
            print("Количество для возврата должно быть положительным.")
            return
        if self.products[product] < quantity:
            print(f"Ошибка: в заказе меньше товара '{product.name}', чем пытаются вернуть.")
            return
        self.products[product] -= quantity
        product.update_stock(quantity)
        if self.products[product] == 0:
            del self.products[product]

class Store:
    def __init__(self):
        self.products = []

    def add_product(self, product: Product):
        self.products.append(product)

    def list_products(self):
        print("Список товаров в магазине:")
        for product in self.products:
            print(f"- {product.name}: Цена = {product.price}, Остаток на складе = {product.stock}")

    def create_order(self):
        return Order()
```

```python
# Создаем магазин и товары
store = Store()
laptop = Product("Ноутбук", 1000, 5)
phone = Product("Смартфон", 500, 10)
headphones = Product("Наушники", 150, 8)

store.add_product(laptop)
store.add_product(phone)
store.add_product(headphones)

# Просмотр доступных товаров
store.list_products()

# Создаем новый заказ
order = store.create_order()

# Добавляем товары в заказ, включая попытку добавить больше товара, чем на складе
order.add_product(laptop, 3)         # успешно
order.add_product(phone, 11)         # ошибка: недостаточно на складе
order.add_product(headphones, 5)     # успешно

# Вычисляем общую стоимость
print("Общая стоимость:", order.calculate_total())

# Удаляем часть товара из заказа
order.remove_product(headphones, 3)  # уменьшаем количество наушников в заказе

# Возвращаем товар в магазин
order.return_product(laptop, 1)      # возвращаем 1 ноутбук

# Итоговая стоимость после изменений
print("Общая стоимость после изменений:", order.calculate_total())

# Проверяем остаток на складе и товары в заказе
store.list_products()
print("Товары в заказе:")
for product, qty in order.products.items():
    print(f"{product.name}: {qty}")

# Попытка удалить товар, которого нет в заказе
order.remove_product(phone, 1)       # ошибка: товар отсутствует в заказе

# Попытка вернуть больше товара, чем есть в заказе
order.return_product(headphones, 10) # ошибка

# Проверка добавления отрицательного количества
order.add_product(phone, -2)         # ошибка
```
