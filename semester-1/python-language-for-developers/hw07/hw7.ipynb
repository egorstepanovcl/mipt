# Задание 1. Получение данных из публичного API

1. Выберите публичный API. Например, [JSONPlaceholder](https://jsonplaceholder.typicode.com/).
2. Напишите скрипт, который:
    - отправляет GET-запрос к `/posts`,
    - извлекает и выводит заголовки и тела первых 5 постов.

```python
import requests

def get_first_5_posts():
    url = "https://jsonplaceholder.typicode.com/posts"
    response = requests.get(url)
    if response.status_code == 200:
        posts = response.json()
        for post in posts[:5]:
            print(f"Заголовок: {post['title']}")
            print(f"Тело поста: {post['body']}\n")
    else:
        print("Ошибка при получении данных:", response.status_code)

get_first_5_posts()
```

# Задание 2. Работа с параметрами запроса

1. Используйте API [OpenWeather](https://openweathermap.org/api) для получения данных о погоде.
2. Напишите программу, которая:
    - принимает название города от пользователя,
    - отправляет GET-запрос к API и выводит текущую температуру и описание погоды.

Для повторной проверки необходимо в Секреты Google Colab добавить API ключ OpenWeatherMap с именем `OPENWEATHER_API_KEY`, если этого не сделать, то API ключ будет необходимо ввести с клавиатуры.

```python
import requests
from google.colab import userdata

def get_weather():
    city = input("Введите название города: ")

    try:
        api_key = userdata.get('OPENWEATHER_API_KEY')
    except (userdata.SecretNotFoundError, userdata.NotebookAccessError):
        api_key = input("API ключа нет в Секретах Google Colab. Введите API ключ: ")

    url = f"http://api.openweathermap.org/data/2.5/weather?q={city}&appid={api_key}&units=metric&lang=ru"

    response = requests.get(url)
    if response.status_code == 200:
        data = response.json()
        temp = data['main']['temp']
        description = data['weather'][0]['description']
        print(f"Температура в городе {city}: {temp}°C")
        print(f"Описание погоды: {description}")
    else:
        print("Ошибка при получении данных о погоде:", response.status_code)

get_weather()
```

# Задание 3. Создание и обработка POST-запросов

1. Выберите API, который позволяет создавать ресурсы. Например, [JSONPlaceholder](https://jsonplaceholder.typicode.com/).
2. Напишите программу, которая:
    - отправляет POST-запрос для создания нового поста,
    - выводит ID созданного поста и его содержимое.

```python
import requests

def create_post(title, body, user_id):
    url = "https://jsonplaceholder.typicode.com/posts"
    post_data = {
        "title": title,
        "body": body,
        "userId": user_id
    }
    response = requests.post(url, json=post_data)
    if response.status_code == 201:
        post = response.json()
        print(f"Создан пост с ID: {post['id']}")
        print(f"Содержимое: {post}")
    else:
        print(f"Ошибка создания поста: {response.status_code}")

if __name__ == "__main__":
    title = "Новый заголовок поста"
    body = "Текст содержимого нового поста"
    user_id = 1
    create_post(title, body, user_id)
```

# Задание 4. Обработка ошибок и работа с данными

1. Расширьте предыдущий код для обработки разных кодов состояния, например: 400, 404.
2. Добавьте вывод сообщения об ошибке в зависимости от полученного кода состояния.

```python
import requests

def create_post(title, body, user_id):
    url = "https://jsonplaceholder.typicode.com/posts"
    post_data = {
        "title": title,
        "body": body,
        "userId": user_id
    }
    response = requests.post(url, json=post_data)

    if response.status_code == 201:
        post = response.json()
        print(f"Создан пост с ID: {post['id']}")
        print(f"Содержимое: {post}")
    elif response.status_code == 400:
        print("Ошибка 400: Неверный запрос - проверьте данные")
    elif response.status_code == 404:
        print("Ошибка 404: Ресурс не найден")
    else:
        print(f"Неизвестная ошибка: Код состояния {response.status_code}")

if __name__ == "__main__":
    title = "Новый заголовок поста"
    body = "Текст содержимого нового поста"
    user_id = 1
    create_post(title, body, user_id)
```
