# Задание 1. Золото в рюкзаке

| Ограничение | Значение |
| ----------- | -------- |
| Ограничение времени | 1 секунда |
| Ограничение памяти | 64 Мб |
| Ввод | input.txt |
| Вывод | output.txt |

Следуя указаниям на карте сокровищ, команда капитана Флинта нашла клад: слитки золота много лет назад были зарыты на необитаемом острове под пальмой.

С собой у команды один, хоть и вместительный, рюкзак, в который надо положить как можно больше золота.

Найдите максимальный вес золота, который можно унести в рюкзаке.

**Формат ввода**

Первая строка содержит целое число $1 \leq W \leq 10^5$ - вместимость рюкзака.

Во второй строке дано число золотых слитков $n$, $1 \leq n \leq 500$.

Третья строка содержит список $n$ целых чисел $0 \leq w_1, \ldots, w_n \leq 10^5$, где $w_i$ - вес $i$-ого слитка.


**Формат вывода**

Выведите одно целое число - максимальный вес золота в слитках, которое можно унести в рюкзаке.

```python
data = '''10
3
1 4 8
'''

with open('input.txt', 'w', encoding='utf-8') as f:
    f.write(data)
```

```python
with open('input.txt', 'r', encoding='utf-8') as f:
    W = int(f.readline().strip())
    n = int(f.readline().strip())
    weights = list(map(int, f.readline().strip().split()))

dp = [0] * (W + 1)

for i in range(n):
    weight = weights[i]
    for w in range(W, weight - 1, -1):
        dp[w] = max(dp[w], dp[w - weight] + weight)

result = dp[W]

with open('output.txt', 'w', encoding='utf-8') as f:
    f.write(str(result))
```

# Задание 2. Назначение ответственных

| Ограничение | Значение |
| ----------- | -------- |
| Ограничение времени | 5 секунд |
| Ограничение памяти | 64 Мб |
| Ввод | input.txt |
| Вывод | output.txt |

Контроль за выполнением задач на участке $[s_i, f_i]$, где $s_i, f_i \in \mathbb{Z}$ (целые числа) может осуществлять ответственное лицо, находясь в точке $x_i \in \mathbb{Z}$, такой что $s_i \leq x_i \leq f_i$. При этом контроль из точки $x_i$ может осуществляться за любым количеством участков, главное, чтобы выполнялось условие, что точка $x_i$ находится внутри отрезка, обозначающего участок.

Известен план работ на $n$ участках, для каждого участка определены его границы.

Требуется найти наименьшее возможное число ответственных за контроль задач на всех указанных участках, т.е. мощность минимального множества $C$, состоящего из целых точек, таких, что каждый из отрезков $[s_i, f_i]$ содержит хотя бы одну из точек этого множества $C$.

**Формат ввода**

В первой строке дано число $1 \leq n \leq 100$ отрезков.

Каждая из последующих $n$ строк содержит по два целых числа $0 \leq s_i \leq f_i \leq 10^9$, задающих начало и конец отрезка.


**Формат вывода**

Выведите одно целое число — минимальное число точек, удовлетворяющих условиям задачи.

```python
data = '''3
1 3
2 5
3 6
'''

with open('input.txt', 'w', encoding='utf-8') as f:
    f.write(data)
```

```python
with open('input.txt', 'r', encoding='utf-8') as file:
    n = int(file.readline().strip())
    intervals = []
    for _ in range(n):
        s, f = map(int, file.readline().strip().split())
        intervals.append((s, f))

intervals.sort(key=lambda x: x[1])

points_count = 0
last_point = -float('inf')

for start, end in intervals:
    if start > last_point:
        points_count += 1
        last_point = end

with open('output.txt', 'w', encoding='utf-8') as file:
    file.write(str(points_count))
```
