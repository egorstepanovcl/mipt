# Задание 1. Регионы мира

| Ограничение | Значение |
| ----------- | -------- |
| Ограничение времени | 1 секунда |
| Ограничение памяти | 64 Мб |
| Ввод | input.txt |
| Вывод | output.txt |

В файле input.txt, содержимое которого представлено в формате csv c разделителем ';' (точка с запятой), перечислены страны. О каждой стране известны ее название, регион, площадь, ВВП, уровень образованности населения (в процентах).

Первая строка файла - заголовки столбцов.

Укажите регион(ы) мира, в которых наиболее грамотное население.

Для этого для каждого региона определите долю стран из заданного списка, в которых грамотность (последний столбец Literacy) выше 90%.

Сравните регионы по вычисленному показателю: чем больше доля "образованных" стран, тем более грамотное население в регионе. В результате должен быть найден регион или регионы, где рассчитанное значение максимально.

Если найденных регионов несколько, выведите их все в алфавитном порядке.

**Формат ввода**

В первой строке записаны заголовки столбцов.

В следующих строка представлена через символ ';' информация о странах: название, регион, площадь, ВВП, уровень образованности населения.

**Формат вывода**

Список регионов в алфавитном порядке, название каждого региона должно быть выведено с новой строки.

Результат должен быть записан в файл output.txt.

```python
data = '''Country;Region;Population;Area;GDP;Literacy
Saint Lucia;LATIN AMER. & CARIB;168458;616;5400.0;67.0
Guinea;SUB-SAHARAN AFRICA;9690222;245857;2100.0;35.9
Paraguay;LATIN AMER. & CARIB;6506464;406750;4700.0;94.0
Burkina Faso;SUB-SAHARAN AFRICA;13902972;274200;1100.0;26.6
Laos;ASIA (EX. NEAR EAST);6368481;236800;1700.0;66.4
Cyprus;NEAR EAST;784301;9250;19200.0;97.6
Mongolia;ASIA (EX. NEAR EAST);2832224;1564116;1800.0;97.8
Indonesia;ASIA (EX. NEAR EAST);245452739;1919440;3200.0;87.9
Switzerland;WESTERN EUROPE;7523934;41290;32700.0;99.0
Uzbekistan;C.W. OF IND. STATES;27307134;447400;1700.0;99.3
'''

with open('input.txt', 'w', encoding='utf-8') as f:
    f.write(data)
```

```python
import csv

with open('input.txt', 'r', encoding='utf-8') as f:
    reader = csv.DictReader(f, delimiter=';')
    data = list(reader)

region_summary = {}

for row in data:
    region = row['Region']
    literacy = float(row['Literacy'])
    if region not in region_summary:
        region_summary[region] = {'total': 0, 'educated': 0}
    region_summary[region]['total'] += 1
    if literacy > 90.0:
        region_summary[region]['educated'] += 1

region_ratios = {}
for region, stats in region_summary.items():
    if stats['total'] > 0:
        ratio = stats['educated'] / stats['total']
        region_ratios[region] = ratio

if region_ratios:
    max_ratio = max(region_ratios.values())
    best_regions = [region for region, ratio in region_ratios.items() if ratio == max_ratio]
    best_regions.sort()

    with open('output.txt', 'w', encoding='utf-8') as fout:
        for region in best_regions:
            fout.write(region + '\n')
```

# Задание 2. Группы анаграмм

| Ограничение | Значение |
| ----------- | -------- |
| Ограничение времени | 1 секунда |
| Ограничение памяти | 64 Мб |
| Ввод | input.txt |
| Вывод | output.txt |

Дан набор слов. Два слова являются анаграммами, если одно можно получить из другого перестановкой букв (игнорируя регистр). Необходимо сгруппировать слова по анаграммам.

**Формат ввода**

Дана одна строка, в которой через пробел записаны слова.

**Формат вывода**

Выведите список групп анаграмм, каждая группа на своей строке, слова в группе разделены пробелами.

При этом учтите, что
- регистр букв сохраняется исходный;
- каждая группа отсортирована лексикографически (с учетом регистра);
- группы отсортированы по первому элементу группы;

```python
data = "eat tea tan ate nat bat"

with open('input.txt', 'w', encoding='utf-8') as f:
    f.write(data)
```

```python
from collections import defaultdict

with open('input.txt', 'r', encoding='utf-8') as fin:
    words = fin.read().split()

groups = defaultdict(list)
for word in words:
    key = ''.join(sorted(word.lower()))
    groups[key].append(word)

result = []
for group in groups.values():
    result.append(sorted(group))

result.sort(key=lambda g: g[0])

with open('output.txt', 'w', encoding='utf-8') as fout:
    for group in result:
        fout.write(' '.join(group) + '\n')
```
